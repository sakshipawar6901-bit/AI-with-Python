```python
import seaborn as sns
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Internal Seaborn Packages

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset('titanic')
df.head()
```

```python
df.shape
```

```python
sns.countplot(x='alive',data=df)
```

```python
sns.countplot(x='sex',data=df,hue='survived',palette='Set2')
```

```python
#KDE Plot
sns.kdeplot(x='age', data=df , color='black')
```

```python
#Distribution plots
sns.displot(x='age' , kde=True ,bins=5 , data=df)
```

```python
sns.displot(x='age',kde=True , bins=5, data=df , hue= df['survived'], palette='Set2')
```

```python
#Scatter Plots
df= sns.load_dataset('iris')
df.head()
```

```python
sns.scatterplot(x='sepal_length', y='petal_length', hue='species' , data=df)
```

```python
sns.scatterplot(x='sepal_width', y='petal_width', hue='species' , data=df)
```

```python
df.shape
```

```python
#Joint plot

sns.jointplot(x='sepal_width', y='petal_width', data=df , hue= 'species')
```

```python
sns.jointplot(x='sepal_width', y='petal_width', data=df , kind= 'reg')
```

```python
#Pairplot
sns.pairplot(df)

```

```python
corr = df.corr(numeric_only=True)
corr
```

```python
#Heatmap
sns.heatmap(corr, annot=True)
```

```python
sns.heatmap(corr, annot=True , cmap=['grey','orange','green'], vmin=-.5 , vmax=0.6)
```
