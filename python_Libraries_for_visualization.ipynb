Basic Graphs

```python
plt.figure(figsize=(5,5))
empid=np.arange(1,10)
salary =np.arange(20,110,10)
plt.bar(empid,salary)
plt.xlabel('Employee_ID')
plt.ylabel('salary')
plt.show()
```

```python
plt.figure(figsize=(8,5))
ax=plt.axes()
ax.set_facecolor('#ECF0F1')
plt.bar(empid,salary,color="#FFA726")
plt.xlabel('Employee_ID')
plt.ylabel('salary')
plt.show()
```

```python
x1=[1,3,5,7]
x2=[2,4,6,8]
y1=[7,7,7,7]
y2=[17,18,29,40]
plt.figure(figsize=(8,5))
ax=plt.axes()
ax.set_facecolor('white')
plt.bar(x1,y1,label='First')
plt.bar(x2,y2,label='Second',color="#F0188E")
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.title('$ Bar $ $ Chart$')
plt.legend()
plt.show()
```

```python
barplot=plt.bar(empid, salary, color='green')
for bar in barplot:
  yval=bar.get_height()
  plt.text(bar.get_x()+bar.get_width()/2.0 , yval , int(yval) , va ='bottom',ha='center')
plt.title('Bar graph example')
plt.xlabel('Employee_ID')
plt.ylabel("salary")
```

Vertical bar graphs

```python
y1=np.array([1,3,5,7,9])
y2=np.array([2,4,6,8,10])
y3=np.array([20,40,60,80,100]) # Corrected: '80.100' was split into '80, 100'
barwidth=0.25
pos1=np.arange(len(y1))
pos2=[x+barwidth for x in pos1]
pos3=[x+barwidth for x in pos2]
#Marking the plot
plt.bar(pos1, y1,  color='#053c40',width=barwidth, label='batting')
plt.bar(pos2, y2, color='#ce0ee3',width=barwidth, label='bowling')
plt.bar(pos3, y3 , color='#a6ba11' ,width=barwidth, label='fielding')
plt.xlabel('Assignee', fontweight='bold')
plt.ylabel('Number of tickets', fontweight='bold')
plt.xticks([i + barwidth for i in range(len(y1))], ['sachin', 'saurav','rahul','dhoni', 'kohli']) # Corrected: Added 'kohli' to match the number of bars
plt.legend()
plt.show()
```

stacked Bar chart

```python
import numpy as np
import matplotlib.pyplot as plt

y1 = np.array([11, 13, 15, 17, 19])
y2 = np.array([21, 41, 61, 81, 101])
y3 = np.array([20, 40, 60, 80, 100])

barwidth = 0.25

pos1 = np.arange(len(y1))
pos2 = [x + barwidth for x in pos1]
pos3 = [x + barwidth for x in pos2]

# Marking the plot
plt.bar(pos1, y1, color='#d62d06', width=barwidth, label='batting')
plt.bar(pos2, y2, color='#f21120', width=barwidth, label='bowling')
plt.bar(pos3, y3, color='#d6d306', width=barwidth, label='fielding')

plt.xlabel('Assignee', fontweight='bold')
plt.ylabel('Number of tickets', fontweight='bold')

plt.xticks(
    [i + barwidth for i in range(len(y1))],
    ['sachin', 'saurav', 'rahul', 'dhoni', 'virat']
)

plt.legend()
plt.show()
```
